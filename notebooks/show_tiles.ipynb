```python
# # Visualize Saved NPY Tiles
# This notebook loads saved tiles from the data preparation step and plots:
# - Raw image
# - Semantic mask
# - Fibre instance mask
# - Axon instance mask
# - Distance transform (fibre)
# - Distance transform (axon)

import numpy as np
import matplotlib.pyplot as plt
import os
from glob import glob
```

```python
# Modify this to your tile path
tile_dir_path = r"D:\pcarrillo\Git_Repos\AimSeg-Monai_3Targets\prepared_data_bf\train_tiles"
assert os.path.exists(tile_dir_path), f"Tile not found: {tile_dir_path}"
all_images =[]
for file in glob(tile_dir_path + "\*.npy"):
    all_images.append(os.path.basename(file))
```

```python
# Plotting utility
def plot_image(ax, img, title, cmap='gray'):
    ax.imshow(img, cmap=cmap)
    ax.set_title(title)
    ax.axis("off")
```

```python
for image_name in all_images:
    # Load ground truth data
    img_filename = os.path.join(tile_dir_path, image_name)

    # Load the tile dictionary
    tile = np.load(img_filename, allow_pickle=True).item()

    # Sanity check: print keys and shapes
    print(image_name)
    for k, v in tile.items():
        print(f"{k}: shape = {v.shape}, dtype = {v.dtype}, min = {v.min()}, max = {v.max()}")

    # Create subplots
    fig, axs = plt.subplots(2, 3, figsize=(15, 10))

    plot_image(axs[0, 0], tile["image"], "Raw Image", cmap=None if tile["image"].ndim == 3 else 'gray')
    plot_image(axs[0, 1], tile["mask_sem"], "Semantic Mask", cmap = "viridis")
    plot_image(axs[0, 2], tile["mask_fibre"], "Fibre Instances", cmap = "viridis")

    plot_image(axs[1, 0], tile["mask_axon"], "Axon Instances", cmap = "viridis")
    plot_image(axs[1, 1], tile["sdt_fibre"], "Skeleton-Aware DT (Fibre)", cmap = "magma")
    plot_image(axs[1, 2], tile["sdt_axon"], "Skeleton-Aware DT (Axon)", cmap = "magma")

    plt.tight_layout()
    plt.show()

```
