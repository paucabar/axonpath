# Inspect prepared tiles

Shows the tiles written by the data preparation step (`scripts/prepare_data.py` or `split_dataset()`), to check the training targets before training. For each tile it prints the arrays it contains and plots:

- the image
- the semantic mask (0 background, 1 fibre, 2 inner cylinder, 3 axon)
- the fibre and axon instance masks
- the fibre and axon skeleton-aware distance transforms (SDTs)

**Usage:** set `tile_dir_path` to a folder of prepared tiles, then run all cells.

```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Folder of prepared .npy tiles (relative to this notebook)
tile_dir_path = Path("../prepared_data/train_tiles")
assert tile_dir_path.is_dir(), f"Tile folder not found: {tile_dir_path}"

tile_paths = sorted(tile_dir_path.glob("*.npy"))
print(f"{len(tile_paths)} tiles found")
```

```python
# Plotting utility
def plot_image(ax, img, title, cmap='gray'):
    ax.imshow(img, cmap=cmap)
    ax.set_title(title)
    ax.axis("off")
```

```python
for tile_path in tile_paths:
    # Each tile is a dictionary of arrays
    tile = np.load(tile_path, allow_pickle=True).item()

    # Sanity check: print keys and shapes
    print(tile_path.name)
    for k, v in tile.items():
        print(f"{k}: shape = {v.shape}, dtype = {v.dtype}, min = {v.min()}, max = {v.max()}")

    # Create subplots
    fig, axs = plt.subplots(2, 3, figsize=(15, 10))

    plot_image(axs[0, 0], tile["image"], "Raw Image", cmap=None if tile["image"].ndim == 3 else 'gray')
    plot_image(axs[0, 1], tile["mask_sem"], "Semantic Mask", cmap = "viridis")
    plot_image(axs[0, 2], tile["mask_fibre"], "Fibre Instances", cmap = "viridis")

    plot_image(axs[1, 0], tile["mask_axon"], "Axon Instances", cmap = "viridis")
    plot_image(axs[1, 1], tile["sdt_fibre"], "Skeleton-Aware DT (Fibre)", cmap = "magma")
    plot_image(axs[1, 2], tile["sdt_axon"], "Skeleton-Aware DT (Axon)", cmap = "magma")

    plt.tight_layout()
    plt.show()
```
