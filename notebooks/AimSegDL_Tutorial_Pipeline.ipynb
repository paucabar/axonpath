# axonpath: End-to-End Pipeline for Axon and Fibre Segmentation

This notebook demonstrates the **complete axonpath workflow**, from data preparation to model export in BioImage.IO format.

You will learn how to:
1. Prepare datasets for training  
2. Train a U-Net segmentation model with axonpath  
3. Evaluate model performance on a test dataset  
4. Run inference on new images  
5. Extract quantitative metrics (g-ratios, morphometrics, spatial metrics)  
6. Export the trained model as a standardized **BioImage.IO package**

## Data Preparation

Before training, the raw datasets must be tiled and split into train/validation/test subsets.
This step also computes auxiliary distance maps required by axonpath.

The function `split_dataset()` expects the dataset root directory to follow this structure:

- datasets/
  - dataset1/
    - images/
    - masks/
    - labels/

Each image in the `images/` folder must have a corresponding mask and label file with the **exact same filename**.

```python
from axonpath.data_preparation.data_preparation import split_dataset
```

```python
# Prepare dataset tiles and splits for training
split_dataset(
    in_root = "datasets_em",
    out_root = "prepared_data_em",
    create_test_split = True,
)
```

## Model Training

The training process uses a configurable `TrainingConfig` object.  
You can adjust parameters such as learning rate, batch size, number of epochs, or paths to pretrained weights.

Training output includes:
- Model checkpoints (`model_checkpoint.pth.tar`)
- Best weights file (`best_weights_model.pth`)
- Last weights file (`last_epoch_model.pth`)
- Val and train loss plots (`loss_plot.pdf`, `loss_plot_log.pdf`)
- Segmentation scores plot (`segmentation_scores_plot`)

```python
import torch
from axonpath.training.config import TrainingConfig
from axonpath.training.train import train
```

```python
# Define training configuration
train_config = TrainingConfig(
    learning_rate = 1e-3,
    batch_size = 8,
    num_epochs = 2000,
    num_workers = 0,
    image_height = 512,
    image_width = 512,
    pin_memory = True,
    device = "cuda" if torch.cuda.is_available() else "cpu",
    fibre_threshold = 0.5,
    axon_threshold = 0.5,
    min_diameter = 30.0,
    train_dir = "prepared_data_em/train_tiles/",
    val_dir = "prepared_data_em/val_tiles/",
    pretrained_weights = None,     # optional: path to .pth weights file or model name (if available) to initialize model
    load_checkpoint = False,       # resume from previous checkpoint if True
    model_name = "axonpath",
    )
```

```python
# Train the model
%matplotlib inline
train(train_config)
```

### Model Evaluation

After training, you can evaluate your model on a test dataset.  
The evaluation computes **F1**, **Precision**, **Recall**, and **Jaccard** for fibres, axons, and inner tongues.

Results are automatically saved as `.tsv` files and can be visualized as summary bar charts.

```python
import torch
from axonpath.evaluation.evaluate_model import evaluate_model_on_testset, compute_imagewise_means, plot_summary_bar
```

```python
test_dir = "prepared_data_em/test_tiles/"
model_path = "best_weights_model.pth"
device = "cuda" if torch.cuda.is_available() else "cpu"
min_diameter = 30.0

%matplotlib inline
fibre_results, axon_results, inner_tongue_results = evaluate_model_on_testset(
    model_path,
    test_dir,
    device,
    min_diameter,
    output_csv = "Evaluation_Results",
    display_figure=True, # Display ground truth and predicted labels
    show_plots = True # Display F1 for 0.5-0.9 IoU threholds
)
```

```python
# Plot overall mean metrics for each structure
mean_metrics_fibre = compute_imagewise_means(fibre_results)
plot_summary_bar(mean_metrics_fibre, "Fibre")

mean_metrics_axon = compute_imagewise_means(axon_results)
plot_summary_bar(mean_metrics_axon, "Axon")

mean_metrics_inner_tongue = compute_imagewise_means(inner_tongue_results)
plot_summary_bar(mean_metrics_inner_tongue, "Inner_Tongue")
```

## Inference on New Images

You can perform inference using pretrained weights (`em`, `bf`) or your own model file.

This step loads a trained model, runs segmentation, and visualizes:
- The input image  
- The semantic segmentation map  
- Fibre instance segmentation  
- Axon instance segmentation

```python
import matplotlib.pyplot as plt
from axonpath.inference import load_image, load_model, run_inference
```

```python
# Example image (choose one depending on model type)
example_image_path = r"axonpath\example_images\P03A_Frame5_roi0.tif"
#example_image_path = r"axonpath\example_images\A72E_roi0.tif"

model_path = "em" # "bf" for brightfield pretrained weights, or custom .pth path

# Load input image and trained model
image = load_image(example_image_path)
model = load_model(model_path)

# Run inference
fibre_threshold=0.5
axon_threshold=0.5
min_diameter=30.0
labels_fibre, labels_axon, semantic = run_inference(
    image=image,
    model=model,
    fibre_threshold=fibre_threshold,
    axon_threshold=axon_threshold,
    min_diameter=min_diameter
)
```

```python
# Visualize results
fig, axs = plt.subplots(1, 4, figsize=(18, 6))

axs[0].imshow(image, cmap='gray')
axs[0].set_title("Input Image")
axs[0].axis('off')

axs[1].imshow(semantic, cmap='viridis')
axs[1].set_title("Semantic Segmentation")
axs[1].axis('off')

axs[2].imshow(labels_fibre, cmap='nipy_spectral', interpolation="nearest")
axs[2].set_title("Fibre Instances")
axs[2].axis('off')

axs[3].imshow(labels_axon, cmap='nipy_spectral', interpolation="nearest")
axs[3].set_title("Axon Instances")
axs[3].axis('off')

plt.tight_layout()
plt.show()
```

## Measure

After inference, axonpath allows quantitative analysis of fibre and axon morphometry.

This step computes per-fibre features such as:
- Area and diameter of fibre, inner tongue, and axon  
- Myelin and axon g-ratios  
- Fibre circularity, solidity, eccentricity  
- Minimum border and centroid distances  
- Local confluence (neighbour density)

```python
from axonpath.measure.axon_metrics import metrics_table
from axonpath.utils.image_processing import fill_labels, map_axon_labels_to_fibres, remove_unmapped_labels
from skimage.measure import label
```

```python
# Postprocess predicted masks
pred_inner_tongue = label(semantic == 2)
pred_inner_tongue = fill_labels(pred_inner_tongue)
mapped_inner_tongue = map_axon_labels_to_fibres(labels_fibre, pred_inner_tongue)
labels_fibre = remove_unmapped_labels(labels_fibre, mapped_inner_tongue)

# Compute metrics table
df = metrics_table(labels_fibre, mapped_inner_tongue, pixel_size_um=0.008, axon_labels=labels_axon)
df.head()
```

## Export as BioImage.IO Model

Finally, export the trained axonpath model in **BioImage.IO** format.  
This makes your model interoperable with the BioImage Model Zoo and compatible with QuPath or Fiji.

Provide model metadata such as:
- Model name, version, and license  
- Pixel size and minimum expected diameter  
- Author names and citation details  
- Path to example test image

```python
from axonpath.export_utils.create_bioimageio_model import export_bioimageio
from axonpath.export_utils.config import BioimageioExportConfig
```

```python
# Configure export metadata
bioimageio_config = BioimageioExportConfig(
    model_path = f"best_weights_model.pth",
    model_name = "axonpath",
    model_version = "0.1.0",
    test_img_path = r"axonpath\example_images\P03A_Frame5_roi0.tif",
    output_dir = "bioimageio_model",
    model_pixel_size = 0.008,
    min_diameter = 30.0,
    predict_inner_cylinder = True,
    citation_text = "Carrillo-Barberà et al., 2025",
    citation_doi = "10.1234/fake-doi-placeholder",
    author_names = ["Pau Carrillo-Barberà", "Thibaut Goldsborough", "Alan O'Callaghan", "Andrea Poveda Sabuco", "Chiara Sgattoni", "Jose Antonio Gómez Sánchez", "Ana Rondelli", "Anna Williams", "Peter Bankhead"],
    license_id = "CC-BY-4.0",
)
```

```python
# Run export
export_bioimageio(bioimageio_config)
```
