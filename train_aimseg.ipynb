# Aimsegdl training pipeline

```python
import torch
from aimsegdl.data_preparation.data_preparation import create_train_val_test_split_all
from aimsegdl.training.config import TrainingConfig
from aimsegdl.training.train import train
```

# Data preparation

```python
create_train_val_test_split_all(
    in_root = "datasets_em",
    out_root = "prepared_data_em",
)
```

# Training

```python
config = TrainingConfig(
    learning_rate = 1e-3,
    batch_size = 8,
    num_epochs = 20,
    num_workers = 0,
    image_height = 512,
    image_width = 512,
    pin_memory = True,
    device = "cuda" if torch.cuda.is_available() else "cpu",
    train_dir = "prepared_data_em/train_tiles/",
    val_dir = "prepared_data_em/val_tiles/",
    pretrained_weights = None,       # .pt weights (preferred entry point)
    load_checkpoint = False,       # Resume from checkpoint
    bioimageio = False,
    model_name = "aimsegdl",
    )
```

```python
train(config)
```
